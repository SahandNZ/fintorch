```python
os.environ.setdefault("FINTORCH_BASE_TIME_FRAME", "60")
os.environ.setdefault("FINTORCH_CANDLES_COUNT", "100000")
```

```python
import numpy as np
from rich.progress import Progress

from fintorch.enum import TimeFrame
from fintorch.exchange import ONLINE_EXCHANGE
from fintorch.setting import RICH_PROGRESS_COLUMNS
from fintorch.utils.args import DefaultArgumentParser
```

```python
string_args = [
    "--time-frame", "60",
    "--stf-config", "btc-1d"
]
args = DefaultArgumentParser.parse(args=string_args)
```

```python
with Progress(*RICH_PROGRESS_COLUMNS) as progress:
        overall_task = progress.add_task("Overall", total=len(args.symbols))
        for symbol in ["BTC-USDT"]:
            ONLINE_EXCHANGE.future.data.update_candles_dataframe(
                symbol=symbol,
                time_frame=args.time_frame,
                progress=progress
            )

            progress.update(task_id=overall_task, advance=1)
```
