```python
import logging.config

from fintorch.utils.args import DefaultArgumentParser
```

```python
string_args = ["--symbols-config", "1"]
args = DefaultArgumentParser.parse(args=string_args)
logging.config.dictConfig(args.logging_config_dict)
```

```python
with args.engine:
    print(args.engine)
    symbols = args.engine.strategy.module.dataset.symbols
    time_frames = args.engine.strategy.module.dataset.time_frames + [args.engine.clock.interval]
    dc = args.engine.exchange.future.data.get_data_collection(symbols=symbols, time_frames=time_frames)
    args.engine.show_positions_plot(dc=dc, start_date=args.start_date, stop_date=args.stop_date)
```
