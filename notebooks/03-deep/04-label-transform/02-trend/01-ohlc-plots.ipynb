```python
from fintorch.deep.transform.label import *
from fintorch.enum import TimeFrame
from fintorch.exchange import ONLINE_EXCHANGE
```

```python
SYMBOL = "BTC-USDT"
TIME_FRAME = TimeFrame.DAY1

START_DATE = "2023-01-01"
STOP_DATE = "2024-06-01"
```

# Load data collection

```python
dc = ONLINE_EXCHANGE.future.data.get_data_collection(symbols=[SYMBOL], time_frames=[TIME_FRAME])
```

# draw ohlc plots

### Forward Backward Min Method

```python
transform = ForwardBackwardMinimumLabelTransform(symbol=SYMBOL, time_frame=TIME_FRAME)
transform.show_ohlc_plot(dc=dc, start_date=START_DATE, stop_date=STOP_DATE)
```

### Forward Ichimoku Method

```python
transform = ForwardIchimokuLabelTransform(symbol=SYMBOL, time_frame=TIME_FRAME)
transform.show_ohlc_plot(dc=dc, start_date=START_DATE, stop_date=STOP_DATE)
```

### Forward Middle SMA Method

```python
transform = ForwardMiddleSmaLabelTransform(symbol=SYMBOL, time_frame=TIME_FRAME)
transform.show_ohlc_plot(dc=dc, start_date=START_DATE, stop_date=STOP_DATE)
```

### Forward ROC Method

```python
transform = ForwardRocLabelTransform(symbol=SYMBOL, time_frame=TIME_FRAME)
transform.show_ohlc_plot(dc=dc, start_date=START_DATE, stop_date=STOP_DATE)
```

### Next Fractal Method

```python
transform = NextFractalLabelTransform(symbol=SYMBOL, time_frame=TIME_FRAME)
transform.show_ohlc_plot(dc=dc, start_date=START_DATE, stop_date=STOP_DATE)
```
