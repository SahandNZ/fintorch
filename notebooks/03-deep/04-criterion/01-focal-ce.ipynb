```python
import torch
import numpy as np 
from torch import nn
import torch.nn.functional as F
```

```python
input = torch.tensor(
    [[0.1, 0.9],
    [0.3, 0.7],
    [0.7, 0.3]]
).float()

target = torch.tensor(
    [[0, 1],
     [1, 0],
     [1, 0]]
).float()

input, target
```

```python
weight = torch.sum(target, dim=0) / len(target)
loss = F.cross_entropy(input=input, target=target, weight=weight, reduction="none")

actual = torch.argmax(target, dim=-1)
p_t = input[range(len(input)), actual]
focal_loss = loss * ((1 - p_t) ** 2)

print(p_t)
print(loss)
print(focal_loss)
```
