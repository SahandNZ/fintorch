```python
import torch
import numpy as np

import fintorch
from fintorch.deep.metrics import Metrics
from fintorch.utils.args import DefaultArgumentParser
```

```python
MODULE_INDEX = 10
START_DATE = "2022-01-01"
STOP_DATE = "2022-03-01"

# arguments
string_args = [
    "--start-date", START_DATE,
    "--stop-date", STOP_DATE,
]
args = DefaultArgumentParser.parse(args=string_args)

# prepare to show plot
dc = args.online_exchange.future.data.get_data_collection(
    symbols=args.module.dataset.symbols,
    time_frames=args.module.dataset.time_frames
)

# show metrics and plots
with args.modules[MODULE_INDEX]:
    metrics = args.module.calculate_metrics(dc=dc, start_date=args.start_date, stop_date=args.stop_date)
    print(metrics)
    args.module.show_ohlc_plot(dc=dc, start_date=args.start_date, stop_date=args.stop_date)
    
```
