```python
from fintorch.utils.args import DefaultArgumentParser
```

```python
string_args = [
    "--start-date", "2024-03-01",
    "--stop-date", "2024-06-01",
]
args = DefaultArgumentParser.parse(args=string_args)
```

```python
dc = args.online_exchange.future.data.get_data_collection(symbols=args.symbols, time_frames=args.time_frames)
```

```python
for label_transform in args.label_transforms:
    label_transform.show_ohlc_plot(dc=dc, start_date=args.start_date, stop_date=args.stop_date)
    print("\n" * 2)
```
