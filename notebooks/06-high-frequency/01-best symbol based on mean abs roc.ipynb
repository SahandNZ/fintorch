```python
import os
import math
import json
import copy
import itertools
import functools
from datetime import datetime
```

```python
import numpy as np
from rich.progress import Progress

from fintorch.enum import TimeFrame
from fintorch.exchange import ONLINE_EXCHANGE
from fintorch.setting import RICH_PROGRESS_COLUMNS
from fintorch.utils.function import call_with_dict
from fintorch.utils.args import DefaultArgumentParser
from fintorch.deep.transform.label import ForwardRocLabelTransform
```

```python
os.environ.setdefault("FINTORCH_BASE_TIME_FRAME", "60")

string_args = [
    "--stf-config", "all-daily",
    "--stop-date", "2024-06-01",
    "--time-frame", "60",
]
args = DefaultArgumentParser.parse(args=string_args)
```

```python
results = []

for symbol in args.symbols:
    df = ONLINE_EXCHANGE.future.data.get_candles_dataframe(symbol=symbol, time_frame=args.time_frame).copy()
    df["roc"] = df.close / df.open - 1
    cdf = df[0.001 <= np.abs(df.roc)]
    value = round(np.abs(cdf.roc).mean(), 4)
    results.append((value, symbol))
```

```python
sorted_results = sorted(results, key=lambda item: item[0], reverse=True)
for index, item in enumerate(sorted_results):
    print(index, item)
```
